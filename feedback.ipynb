```python
def write_feedback():
    feedback = input("Write your Feedback: ")

    good = ["good", "great", "excellent", "awesome", "fantastic","outstanding", "super", "nice", "wonderful", "perfect"]

    bad = ["bad", "worst", "poor", "awful","disappointing", "unsatisfactory"]

    try:
        if any(word in feedback.lower() for word in good):
            with open("good_feedback.txt", "a") as file:
                file.write(feedback + "\n")

        elif any(word in feedback.lower() for word in bad):
            with open("bad_feedback.txt", "a") as file:
                file.write(feedback + "\n")

        else:
            with open("natural_feedback.txt", "a") as file:
                file.write(feedback + "\n")

    except Exception as e:
        print("Error occurred while saving feedback.")

    else:
        print("Thank you for your valuable feedback!!!!")

    finally:
        print("Feedback submission successfully completed")


def read_feedback():
    try:
        with open("good_feedback.txt", "r") as good_file:
            good_feedback = good_file.readlines()

        with open("bad_feedback.txt", "r") as bad_file:
            bad_feedback = bad_file.readlines()

        with open("natural_feedback.txt", "r") as natural_file:
            natural_feedback = natural_file.readlines()

        print("Good Feedback:")
        for feedback in good_feedback:
            print(feedback.strip())

        print("Bad Feedback:")
        for feedback in bad_feedback:
            print(feedback.strip())

        print("Natural Feedback:")
        for feedback in natural_feedback:
            print(feedback.strip())

    except FileNotFoundError:
        print("File not found")

    finally:
        print("Thank you for your valuable feedback")


print("Code run successfully...")
```
