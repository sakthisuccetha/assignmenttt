```python
n = int(input("Enter side length: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
n = int(input("Enter height: "))

for i in range(1, n + 1):
    # Spaces before stars
    for j in range(n - i):
        print(" ", end=" ")

    # Stars
    for j in range(2 * i - 1):
        if j == 0 or j == 2 * i - 2 or i == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```

```python
n = int(input("Enter height: "))

# Upper half
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end=" ")

    for j in range(2 * i - 1):
        if j == 0 or j == 2 * i - 2:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()

# Lower half
for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end=" ")

    for j in range(2 * i - 1):
        if j == 0 or j == 2 * i - 2:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```

```python
n = int(input("Enter odd grid size: "))

middle = n // 2

for i in range(n):
    for j in range(n):
        if i == middle or j == middle:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
n = int(input("Enter grid size: "))

for i in range(n):
    for j in range(n):
        if j == i or j == n - i - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```
