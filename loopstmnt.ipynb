```python

total = 0

for i in range(1, 101):
    total = total + i

print("Sum =", total)
```

```python

n = int(input("Enter a number: "))

i = 1
while i <= 10:
    print(n, "x", i, "=", n * i)
    i += 1
```

```python

numbers = [10, 15, 22, 33, 40, 51, 64]

even = 0
odd = 0

for num in numbers:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
#reverse a number

num = int(input("Enter a number: "))
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number:", reverse)
```

```python
#prime number 

num = int(input("Enter a number: "))

if num < 2:
    print("Not a prime number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
```

```python
#star pattern

for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
```

```python
#factorial using while

num = int(input("Enter a number: "))

if num < 0:
    print("Factorial is not defined for negative numbers")
else:
    factorial = 1
    i = 1

    while i <= num:
        factorial = factorial * i
        i += 1

    print("Factorial =", factorial)
```

```python
#skips

for i in range(1, 21):
    if i % 3 == 0:
        continue
    print(i, end=" ")
```

```python

numbers = [10, 25, 40, 65, 80, 90]

for num in numbers:
    if num > 50:
        print("First number greater than 50:", num)
        break
```

```python
#fibonacci

n = int(input("Enter the number of terms: "))

a = 0
b = 1

if n <= 0:
    print("Enter a positive number")
else:
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b
```
